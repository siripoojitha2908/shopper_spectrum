# **Project Summary -**

Shopper Spectrum is an e-commerce analytics project designed to analyze customer purchasing behavior and provide intelligent business insights through customer segmentation and product recommendations. The project utilizes transaction data from an online retail business to identify meaningful customer groups based on their purchasing patterns and recommend relevant products to customers.

The project begins with data preprocessing, where missing values, duplicate records, cancelled transactions, and invalid entries are removed to ensure data quality. Exploratory Data Analysis (EDA) is then performed to understand customer behavior, sales trends, product performance, and geographical distribution of transactions.

To segment customers, Recency, Frequency, and Monetary (RFM) analysis is applied. These metrics help measure how recently a customer purchased, how often they purchase, and how much they spend. Using these features, customers are grouped into segments such as High-Value Customers, Regular Customers, Occasional Shoppers, and At-Risk Customers through clustering techniques.

In addition to customer segmentation, the project includes a product recommendation system based on Item-Based Collaborative Filtering. Product similarities are calculated using customer purchase history, allowing the system to recommend relevant products similar to a selected item.

The final solution is deployed as a Streamlit web application that provides an interactive interface for customer segmentation and product recommendation. The application enables users to enter customer details to identify their segment and receive product recommendations in real time.

````python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(rfm)
````

````python
# Chart - 12 visualization code
# Elbow Curve
inertia = []

for k in range(2,11):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(rfm_scaled)
    inertia.append(km.inertia_)

plt.plot(range(2,11), inertia, marker='o')

plt.title('Elbow Curve')

plt.xlabel('Clusters')
plt.ylabel('Inertia')

plt.show()
````

##### 1. Why did you pick the specific chart?

The elbow curve is a standard method for determining the optimal number of clusters in K-Means clustering. It helps avoid under-segmentation or over-segmentation.

##### 2. What is/are the insight(s) found from the chart?

The curve shows a point where the reduction in inertia slows down significantly. This point indicates the optimal number of customer segments.

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import f_oneway

cluster0 = rfm[rfm['Cluster'] == 0]['Monetary']
cluster1 = rfm[rfm['Cluster'] == 1]['Monetary']
cluster2 = rfm[rfm['Cluster'] == 2]['Monetary']
cluster3 = rfm[rfm['Cluster'] == 3]['Monetary']

f_stat, p_value = f_oneway(
    cluster0,
    cluster1,
    cluster2,
    cluster3
)

print("F-statistic:", f_stat)
print("P-value:", p_value)
````

## ***7. ML Model Implementation***

### ML Model - 1

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

kmeans = KMeans(n_clusters=4, random_state=42)

kmeans_labels = kmeans.fit_predict(rfm_scaled)

kmeans_score = silhouette_score(rfm_scaled, kmeans_labels)

print("Silhouette Score:", kmeans_score)
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

````python
# Visualizing evaluation Metric Score chart
import matplotlib.pyplot as plt

scores = [kmeans_score]

plt.figure(figsize=(5,4))

plt.bar(['KMeans'], scores)

plt.ylabel("Silhouette Score")

plt.title("KMeans Performance")

plt.show()
````

KMeans Clustering was used to segment customers based on their Recency, Frequency, and Monetary values. The algorithm groups customers into clusters by minimizing the distance between data points and cluster centroids. The model performance was evaluated using the Silhouette Score. A higher Silhouette Score indicates better cluster separation and compactness.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 1 Implementation

# Fit the Algorithm

# Predict on the model
silhouette_scores = []

for k in range(2,11):

    model = KMeans(
        n_clusters=k,
        random_state=42
    )

    labels = model.fit_predict(rfm_scaled)

    score = silhouette_score(
        rfm_scaled,
        labels
    )

    silhouette_scores.append(score)

best_k = range(2,11)[np.argmax(silhouette_scores)]

print("Best K:", best_k)
````

````python
# Hyperparameter Optimization Chart
plt.figure(figsize=(8,5))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker='o'
)

plt.xlabel("Number of Clusters")

plt.ylabel("Silhouette Score")

plt.title("Hyperparameter Tuning for KMeans")

plt.show()
````

##### Which hyperparameter optimization technique have you used and why?

The Elbow Method and Silhouette Score analysis were used for hyperparameter tuning. The number of clusters (K) was varied from 2 to 10, and the value that produced the highest Silhouette Score was selected because it provides better cluster separation and interpretation.

##### Have you seen any improvement? Note down the improvement with updates Evaluation metric Score Chart.

Yes. By selecting the optimal number of clusters based on Silhouette Score, the clustering quality improved. The final model produced more meaningful customer segments with better separation between clusters.

### ML Model - 2

````python
#DBSCAN
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.8,
    min_samples=5
)

dbscan_labels = dbscan.fit_predict(
    rfm_scaled
)

dbscan_score = silhouette_score(
    rfm_scaled,
    dbscan_labels
)

print(
    "Silhouette Score:",
    dbscan_score
)
````

#### 1. Explain the ML Model used and it's performance using Evaluation metric Score Chart.

````python
# Visualizing evaluation Metric Score chart
plt.figure(figsize=(5,4))

plt.bar(
    ['DBSCAN'],
    [dbscan_score]
)

plt.ylabel(
    "Silhouette Score"
)

plt.title(
    "DBSCAN Performance"
)

plt.show()
````

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) groups customers based on density. It can identify clusters of varying shapes and detect noise points. Unlike KMeans, it does not require the number of clusters to be specified in advance

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 2 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

# Fit the Algorithm

# Predict on the model
eps_values = [0.5,0.7,0.8,1.0]

scores = []

for eps in eps_values:

    model = DBSCAN(
        eps=eps,
        min_samples=5
    )

    labels = model.fit_predict(
        rfm_scaled
    )

    if len(set(labels)) > 1:

        score = silhouette_score(
            rfm_scaled,
            labels
        )
        print(
            f"eps={eps}, Silhouette Score={score}"
        )
        scores.append(score)

    else:
        scores.append(0)
````

````python
#chart
plt.figure(figsize=(8,5))

plt.plot(
    eps_values,
    scores,
    marker='o'
)

plt.title(
    "DBSCAN Hyperparameter Tuning"
)

plt.xlabel("eps")

plt.ylabel("Silhouette Score")

plt.show()
````

##### Which hyperparameter optimization technique have you used and why?

Grid-based parameter testing was performed by varying the epsilon value. The parameter that generated the best Silhouette Score was selected.

##### Have you seen any improvement? Note down the improvement with updates Evaluation metric Score Chart.

Yes. After tuning the DBSCAN hyperparameters, particularly the epsilon (eps) value, the clustering performance improved. The optimized model produced better cluster separation and more meaningful customer groups. This improvement is reflected in the increased Silhouette Score observed after tuning. The updated evaluation metric chart shows that the tuned DBSCAN model achieved better clustering quality compared to the initial model.
